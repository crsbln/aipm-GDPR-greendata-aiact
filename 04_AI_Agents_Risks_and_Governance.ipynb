# AI Agents: Risks, Governance, and Sustainable Use

An AI agent is more than a chat response. It can combine a model with
instructions, context, memory, tools, and proposed actions. That makes
it useful, but it also changes the data, security, sustainability, and
oversight questions a project team must ask.

This notebook is theoretical. It does not ask you to connect an agent
to a repository or external system. Use only fictional scenarios and
never paste real personal, confidential, or sensitive data into an AI
tool during this activity.

  affect cost and environmental impact.

- Recognise prompt injection, unsupported claims, and automation bias
  as operational risks.

## 1. What Makes an Agent Different?

A prompt-only assistant returns text. An agent may also retrieve
context, keep state, call tools, and propose or execute actions.

| Capability | Example | Main question for a project team |
|---|---|---|
| Context | Reads a document, ticket, or database record. | Which data enters the context window? |
| Memory | Stores a preference or previous summary. | Where is it stored and for how long? |
| Tool use | Searches a system or creates a draft. | What permission and scope does the tool have? |
| Action | Deletes a file, updates a record, or sends a message. | Who approves an irreversible or high-impact action? |

The model is only one component. The surrounding permissions and data
flow determine much of the real-world risk.

*An agent may receive documents, data stores, cloud context, and tools.
A human approval boundary should sit before consequential actions.*

## 2. Data Can Cross Organisational Boundaries

When someone pastes information into an external AI service, it may be
transmitted to a provider, a processor, a subprocessor, or a region
outside the organisation. Whether information is processed or stored
in the United States, China, the EEA, or another location depends on
the provider, contract, account configuration, selected model, support
operations, and subprocessors. Never infer the answer from a model
name alone.

For personal data, a team should verify:

1. What categories of data enter prompts, files, retrieval, tool
   results, logs, or memory?
2. Whether the provider uses data for training, service improvement,
   or support, under the relevant plan and contract.
3. Where processing, storage, and support access can occur.
4. Whether a transfer outside the EEA occurs and which GDPR Chapter V
   mechanism and processor agreement are relevant.
5. The retention, deletion, access-control, and incident processes.

Data minimisation still applies: do not send an entire document when a
short, redacted extract answers the task.

## 3. GDPR Risks Around Agent Use

| Agent component | GDPR and governance question |
|---|---|
| Prompt or attachment | Is every field necessary for the task? Does it include special-category data, credentials, or confidential information? |
| Retrieval context | Can stale, excessive, or cross-user records be retrieved? |
| Conversation or memory | Is it retained, searchable, or reused beyond the original purpose? |
| Tool output | Does the output expose information to a user who should not see it? |
| Log | Does logging create a new dataset containing personal or sensitive information? |

A model answer is not a lawful basis, a DPIA, or legal advice. It can
help draft questions, but a responsible person must verify the facts,
contracts, and legal interpretation.

## 4. Sustainable Agent Use

There is no universal ranking that makes one model automatically
sustainable in every task. Impact depends on model capability, token
volume, context length, output length, hardware, energy mix, retries,
tool calls, and how often the workflow runs.

Practical questions include:

- Can a smaller or simpler model meet the quality requirement?
- Can the prompt include a short relevant extract instead of a full
  history or document collection?
- Can results be cached rather than regenerated?
- Are retries and tool-call chains capped?
- Does the workflow have a stop condition and a useful success metric?
- Does moving data or workload conflict with residency, latency,
  contract, or availability requirements?

## 5. Tool Use Can Cause Real-World Harm

An agent with write access can do more than make a factual error.

| Situation | What can go wrong | Safer boundary |
|---|---|---|
| Cleans a shared folder | Deletes useful files or evidence. | Read-only inspection; explicit approval before deletion. |
| Reads a web page or document | Follows hidden or malicious instructions. | Treat external content as untrusted data; validate proposed actions. |
| Drafts a customer reply | Sends an inaccurate or harmful message. | Draft only; a person sends it. |
| Updates a record | Changes data without the right authority. | Limit scope and require a named approver. |
| Loops through tools | Spends budget, consumes resources, or creates duplicate work. | Set call, retry, time, and cost limits. |

Prompt injection is a special concern. Instructions can be hidden in a
web page, document, issue, email, or tool output. They are data to
inspect, not instructions the agent should obey.

## 6. Human Oversight Is a Design Choice

Human oversight is more useful when it is concrete.

| Agent capability | Suggested boundary |
|---|---|
| Read public or low-risk material | Allow with a clear source and scope. |
| Draft a summary or policy note | Allow, then review sources and assumptions. |
| Use confidential or personal data | Require an approved data path and documented purpose. |
| Send, delete, publish, spend money, or change a record | Require explicit human approval at the action point. |
| Make an employment, credit, education, health, or other high-impact decision | Do not delegate the decision to the agent. |

Logging should record enough structured information to review what
happened without copying unnecessary personal data into another log.

## 7. EU AI Act and AI Literacy

The EU AI Act's AI literacy obligation is a reason to teach these
limits, not just how to get useful output. People who operate or use
AI systems should understand the system's context, their own
experience, the people affected, and the relevant risks.

For an agent use case, ask:

- Is this only a drafting aid, or does it influence a high-impact
  decision?
- Who is the provider, deployer, operator, and affected person?
- What human oversight, logging, transparency, and incident process
  are needed?
- Which claim is known from evidence, which is an assumption, and
  which needs legal or technical confirmation?

### Why this matters
The EU AI Act introduces a **risk-based approach**: the higher the risk to fundamental rights or safety, the stricter the obligations. Understanding these categories is crucial for compliance planning. Many systems will be "minimal risk" (e.g., spam filters), while high-risk systems (e.g., hiring, medical, credit) face heavier obligations.

> Educational material only. Always verify requirements against the latest consolidated text, guidance, and delegated acts.

## Timeline
- **1 August 2024:** the AI Act entered into force.
- **2 February 2025:** the prohibitions in Article 5 and the AI literacy obligation started applying.
- **2 August 2025:** governance rules and the first general-purpose AI (GPAI) model obligations started applying.
- **2 August 2026:** the main body of the Act starts applying, including most transparency duties and most Annex III high-risk obligations.
- **2 August 2027:** the extra transition for Annex I regulated-product high-risk AI systems ends.

These are the application dates in the Regulation and Commission materials cited in the README; always check later updates and guidance.

## Risk ladder
- **Prohibited practices (Article 5):** banned uses such as certain manipulative, exploitative, social-scoring, and workplace/education emotion-recognition practices.
- **High-risk AI systems (Article 6 + Annex I / Annex III):** systems that trigger stricter lifecycle duties because of their sector and impact.
- **Transparency duties (Article 50):** disclosure duties for specific systems such as chatbots, certain emotion-recognition systems, and deepfakes.
- **GPAI model obligations:** duties for some general-purpose AI model providers sit alongside, not inside, the high-risk system rules.
- **Minimal-risk systems:** most other uses remain largely outside the mandatory requirements.

"Limited risk" is a helpful classroom shorthand, but the Act is structured around separate regimes rather than one single middle bucket.

## Roles and enforcement
- **Core roles:** Provider (develops or places a system/model on the market), Deployer (uses it in operations), Importer, Distributor, and authorised representative.
- **Provider duties (common examples):** risk management, data governance, technical documentation, logging by design, instructions for use, conformity steps, and post-market monitoring when applicable.
- **Deployer duties (common examples):** use the system according to instructions, keep operational oversight, retain logs when under their control, monitor incidents, and complete a DPIA or fundamental-rights impact assessment where the law requires it.
- **Transparency is separate from high-risk:** some systems trigger transparency duties without being high-risk, while others are high-risk or prohibited.

### Penalty ceilings
Under Article 99, the main administrative-fine tiers are:

- up to **EUR 35 million**, or, for an undertaking, up to **7%** of its total worldwide annual turnover from the preceding financial year, whichever is higher, for prohibited-practice infringements under Article 5;
- up to **EUR 15 million**, or, for an undertaking, up to **3%** of that turnover, whichever is higher, for listed operator and notified-body obligations, including many duties governing high-risk systems; and
- up to **EUR 7.5 million**, or, for an undertaking, up to **1%** of that turnover, whichever is higher, for supplying incorrect, incomplete, or misleading information to notified bodies or national competent authorities in reply to a request.

The ceiling is different for smaller organizations: for **SMEs, including start-ups**, Article 99 uses the lower of the fixed amount and percentage. Regulation (EU) 2026/1744 also applies the lower ceiling to **small mid-cap enterprises** for the EUR 15 million/3% and EUR 7.5 million/1% tiers. These are maximum ceilings; authorities consider the circumstances of each infringement when setting a fine.

Providers of general-purpose AI (GPAI) models have a separate Commission enforcement regime under **Article 101**, with fines up to **EUR 15 million or 3%** of worldwide annual turnover from the preceding financial year, whichever is higher, for intentional or negligent infringements listed there.

Sources: [AI Act Article 99 — fines for operators and other persons](https://eur-lex.europa.eu/eli/reg/2024/1689/art_99/oj), [AI Act Article 101 — fines for GPAI providers](https://eur-lex.europa.eu/eli/reg/2024/1689/art_101/oj), and [Regulation (EU) 2026/1744 — 2026 amendments](https://eur-lex.europa.eu/eli/reg/2026/1744/oj).

### Worked classification example
The following example classifies a hiring-ranking system under Annex III and separates provider duties from deployer duties. Use it as a model for the independent mini-case later in the notebook.

```python
scenario = {
    "name": "AI system for candidate ranking in hiring",
    "provider": "HR-Tech-Solutions Inc.",
    "deployer": "Global Corp",
    "risk_bucket": "high-risk",
    "why": "employment use case (Annex III)",
    "provider_duties": [
        "build the risk-management, documentation, logging, and validation controls",
        "supply instructions for use and performance limits to deployers",
        "maintain post-market monitoring and serious-incident processes",
    ],
    "deployer_duties": [
        "use the system under the provider's instructions and keep human review in the loop",
        "monitor outcomes, keep logs under its control, and escalate incidents",
        "complete local GDPR / employment-law assessments before rollout",
    ],
}
print("--- System Classification Card ---")
for key, value in scenario.items():
    if isinstance(value, list):
        print(f"{key.upper()}:")
        for item in value:
            print(f"  - {item}")
    else:
        print(f"{key.upper()}: {value}")
```

```python
# Data Governance: Check for bias/imbalance
# Expanded dataset
training_data = [
    {"id": 1, "gender": "M", "hired": True},
    {"id": 2, "gender": "M", "hired": False},
    {"id": 3, "gender": "M", "hired": True},
    {"id": 4, "gender": "F", "hired": False},
    {"id": 5, "gender": "M", "hired": True},
    {"id": 6, "gender": "F", "hired": True},
    {"id": 7, "gender": "M", "hired": False},
    {"id": 8, "gender": "NB", "hired": False},
]


def check_representativeness(data, sensitive_attr):
    counts = {}
    for row in data:
        val = row.get(sensitive_attr, "Unknown")
        counts[val] = counts.get(val, 0) + 1

    total = len(data)
    print(f"--- Data Governance Audit: {sensitive_attr} ---")
    print(f"Total Records: {total}")

    for k, v in counts.items():
        pct = v / total
        # Using ASCII block for visualization
        bar = "|" * int(pct * 20)
        print(f"  {k:<10} | {bar:<20} | {pct:.1%}")

    # Simple heuristic for warning
    if max(counts.values()) / total > 0.7:
        print(
            "\n[WARNING] Significant imbalance detected. One group dominates >70% of data."
        )
        print("   Action required: Collect more diverse data or apply re-sampling.")
    else:
        print("\n[OK] Balance looks acceptable for this small sample.")


check_representativeness(training_data, "gender")
```

**What this code does:** Runs a simple representativeness/imbalance check by counting how many samples fall into each group (here: `gender`).
It prints a bar chart-like view and warns if one group dominates more than 70% of the dataset (a very rough heuristic for Article 10-style data governance checks).

```python
# Article 5 screening aid — this is not a legal classification tool.
def screen_prohibited_practice_signals(features: list[str]) -> str:
    review_signals = {
        "manipulative_or_subliminal_techniques": (
            "Art 5(1)(a) — manipulative, deceptive, or subliminal techniques "
            "that materially distort behavior and cause, or are reasonably "
            "likely to cause, significant harm"
        ),
        "exploitation_of_vulnerability": (
            "Art 5(1)(b) — exploiting vulnerability linked to age, disability, "
            "or a specific social or economic situation, with materially "
            "distorted behavior and significant-harm conditions"
        ),
        "social_scoring": (
            "Art 5(1)(c) — social scoring that leads to the detrimental or "
            "unfavourable treatment described in the Article"
        ),
        "individual_criminal_risk_from_profiling_only": (
            "Art 5(1)(d) — assessing or predicting an individual's criminal-offence "
            "risk solely from profiling or personality traits, subject to the "
            "Article's supporting-assessment boundary"
        ),
        "untargeted_facial_image_scraping": (
            "Art 5(1)(e) — untargeted scraping of facial images from the internet "
            "or CCTV to create or expand facial-recognition databases"
        ),
        "emotion_recognition_workplace_education": (
            "Art 5(1)(f) — emotion inference in workplaces or education "
            "institutions, subject to the medical or safety exception"
        ),
        "sensitive_biometric_categorization": (
            "Art 5(1)(g) — biometric categorization that infers protected or "
            "sensitive attributes, subject to the narrow labeling/filtering boundary"
        ),
        "real_time_remote_biometric_id_law_enforcement": (
            "Art 5(1)(h) — real-time remote biometric identification in publicly "
            "accessible spaces for law enforcement, subject to narrowly defined "
            "objectives, necessity, authorization, and safeguard requirements"
        ),
    }

    print(f"Screening declared features: {features}")
    matches = [
        review_signals[feature] for feature in features if feature in review_signals
    ]

    if matches:
        print()
        print("[REVIEW] Potential Article 5 signals found:")
        for match in matches:
            print(f"  - {match}")
        print(
            "Check the intended purpose, actual operation, affected people, and exceptions."
        )
        return "REQUIRES_LEGAL_REVIEW"

    print()
    print("[NO SIGNAL FOUND] The keyword screen found no listed Article 5 signal.")
    print("This does not prove that the system is permitted or compliant.")
    return "NO_SIGNAL_FOUND"


screen_prohibited_practice_signals(
    [
        "candidate_ranking",
        "emotion_recognition_workplace_education",
        "automated_interview",
    ]
)
```

**What this code does:** Screens declared system features for signals that require an Article 5 review. It deliberately returns `REQUIRES_LEGAL_REVIEW` rather than declaring a system prohibited, because several Article 5 rules include cumulative conditions, narrow exceptions, or safeguards.

A keyword match is only a triage signal. A real assessment must examine the intended purpose, how the feature actually works, who uses it, who is affected, the likely harm, and whether a statutory exception applies. Likewise, `NO_SIGNAL_FOUND` does not prove that a system is permitted, low-risk, or compliant.

Source: [AI Act Article 5 — prohibited AI practices](https://eur-lex.europa.eu/eli/reg/2024/1689/art_5/oj).

## Logging example
Trace decisions to support post-market monitoring and incident investigation.

```python
# Logging example
def log_decision(
    user_id,
    input_summary,
    model_version,
    output,
    store,
    event_time: datetime | None = None,
):
    if event_time is None:
        event_time = AI_ACT_DEMO_TIME + timedelta(minutes=len(store) * 5)
    timestamp = event_time.isoformat().replace("+00:00", "Z")
    entry = {
        "timestamp": timestamp,
        "event_id": f"evt-{len(store) + 1:04d}",
        "user_id": user_id,
        "input_summary": input_summary,
        "model_version": model_version,
        "output": output,
        "status": "success",
    }
    store.append(entry)


audit_log = []
log_decision(
    "cand-123",
    {"cv_length": 2, "keywords": 15},
    "v1.2.0",
    {"score": 0.71, "rank": "B"},
    audit_log,
    event_time=AI_ACT_DEMO_TIME,
)
log_decision(
    "cand-124",
    {"cv_length": 5, "keywords": 40},
    "v1.2.0",
    {"score": 0.92, "rank": "A"},
    audit_log,
    event_time=AI_ACT_DEMO_TIME + timedelta(minutes=5),
)
log_decision(
    "cand-125",
    {"cv_length": 1, "keywords": 5},
    "v1.2.0",
    {"score": 0.33, "rank": "C"},
    audit_log,
    event_time=AI_ACT_DEMO_TIME + timedelta(minutes=10),
)

print(f"--- Audit Log (Total Entries: {len(audit_log)}) ---")
print(json.dumps(audit_log, indent=2))
```

### Solution 2: Logging with a retention policy

We extend the log function to remove old entries after a defined retention period.

```python
def log_decision_with_retention(
    user_id,
    input_summary,
    model_version,
    output,
    store,
    retention_days=30,
    event_time: datetime | None = None,
    reference_time: datetime | None = None,
):
    event_time = event_time or AI_ACT_DEMO_TIME
    reference_time = reference_time or event_time
    entry = {
        "timestamp": event_time.isoformat().replace("+00:00", "Z"),
        "user_id": user_id,
        "input": input_summary,
        "model_version": model_version,
        "output": output,
    }
    store.append(entry)

    cutoff = reference_time - timedelta(days=retention_days)
    original_count = len(store)
    store[:] = [
        log for log in store if datetime.fromisoformat(log["timestamp"]) > cutoff
    ]
    removed = original_count - len(store)

    if removed > 0:
        print(f"[MAINTENANCE] Removed {removed} old log entries.")


# Test it
audit_store = []
reference_time = AI_ACT_DEMO_TIME + timedelta(days=45)
old_timestamp = (reference_time - timedelta(days=40)).isoformat().replace("+00:00", "Z")
audit_store.append({"timestamp": old_timestamp, "data": "old"})

log_decision_with_retention(
    "user-new",
    "test",
    "v1",
    "result",
    audit_store,
    retention_days=30,
    event_time=reference_time,
    reference_time=reference_time,
)
print(f"Current log size: {len(audit_store)}")
```

### Solution 3: Technical documentation outline

For a high-risk system, Article 11 requires detailed documentation. A minimal outline includes:

1.  **System description:**
    *   Intended purpose (what does it do?).
    *   Intended users (who uses it?).
    *   Hardware/software requirements.
2.  **Development process:**
    *   **Data:** Source of training data, bias checks performed, data lineage.
    *   **Model:** Architecture choice, training methodology, validation metrics (accuracy, F1-score).
3.  **Risk management:**
    *   List of foreseeable risks (e.g., bias against non-native speakers).
    *   Mitigation measures (e.g., human-in-the-loop review).
4.  **Monitoring:**
    *   How the system is logged.
    *   Post-market monitoring plan.

### Solution 4: GDPR vs. the AI Act

The two laws are related, but they regulate different things. A simple way to remember it is:

- **GDPR asks:** are you allowed to process this person's data, and are you doing it fairly?
- **AI Act asks:** is this AI system allowed in the EU, what risk level does it fall into, and what controls must surround it?

| Topic | GDPR | EU AI Act |
| --- | --- | --- |
| Main focus | Protects **personal data** and the rights of identifiable people. | Regulates **AI systems and GPAI models** based on risk to safety and fundamental rights. |
| Trigger | Applies when you process **personal data**. | Applies when you develop, place on the market, put into service, or use covered AI systems/models in scope. |
| Core question | \"Do we have a lawful, fair, transparent basis to use this data?\" | \"Is this AI use prohibited, high-risk, or subject to transparency/documentation duties?\" |
| Main actors | Controller, processor, data subject, DPO. | Provider, deployer, importer, distributor, authorised representative. |
| Main duties | Lawful basis, transparency, minimization, accuracy, retention limits, security, rights handling, DPIA where required. | Risk management, data governance, technical documentation, logging, human oversight, accuracy/robustness, instructions for use, post-market monitoring. |
| What it protects | Privacy, data protection, and individual control over personal data. | Safe and trustworthy AI use, especially where AI can affect people in high-impact contexts. |
| Typical enforcement concern | Unlawful collection, excessive data use, weak security, ignored rights requests. | Prohibited AI practices, missing conformity/documentation steps, weak oversight, poor risk controls. |

### Where they overlap in practice

- **Personal data in AI:** If the AI system uses personal data for training, testing, or inference, GDPR still applies even if the AI Act also applies.
- **Human review:** GDPR Article 22 protects people from certain solely automated significant decisions; the AI Act adds **human oversight** design and operational duties.
- **Risk assessments:** A high-risk AI system often also raises GDPR impact questions, so you may need both AI Act documentation and a **DPIA** under GDPR Article 35.
- **Transparency:** GDPR requires clear information about personal-data processing, while the AI Act requires disclosures and instructions tied to the AI system itself.

In short: **GDPR governs the data**, while the **AI Act governs the AI system and its risk controls**. Many real-world products need to satisfy both at the same time.

### Solution 5: Release-decision quality gate

A strong submission does not need to reach a particular decision, but it should pass all of these checks:

- **Purpose:** describes a specific intended use, affected people, and decision supported by the system.
- **Classification:** cites the relevant Article or Annex and records uncertainty instead of forcing a category.
- **Role:** identifies whether the organization is acting as provider, deployer, importer, distributor, or more than one role.
- **Evidence:** separates verified facts from assumptions and names the documents or tests still missing.
- **Controls:** defines concrete conditions such as human oversight, logging, documentation, monitoring, transparency, or a prohibited-feature redesign.
- **Decision:** chooses `GO`, `PILOT`, or `HOLD`, names an accountable owner, and states what evidence would change the decision.

A polished but unsupported `GO` does not pass this gate. A well-evidenced `HOLD` or constrained `PILOT` can be the stronger PM decision.
